```python
Q1. # A college stores students' marks in a list. Write a function that accepts a list of marks and displays the highest mark, lowest mark, average mark, and number of students who passed (marks of 40 or above).

def analyze_marks(marks):
    highest = max(marks)
    lowest = min(marks)
    average = sum(marks) / len(marks)
    passed_student = sum(1 for mark in marks if mark >= 40)
    
    print(f'Highest marks: {highest}')
    print(f'lowest marks: {lowest}')
    print(f'average marks: {average}')
    print(f'Total_passed_student: {passed_student}')
student_marks = [78,85,89,61,22,15,14]
analyze_marks(student_marks)
```

Q2. An online store maintains a list of mobile prices. Write a function 
to display all mobiles costing more than ₹20,000, calculate the 
total inventory value, and find the most expensive mobile price.

```python
list_of_mobile_price = [15000,25000,69000,45000,63000,9999,89999,149999,159999]
def mobile_store_management(list_of_mobile_price):
    price_more_than_20000 = [price for price in list_of_mobile_price if price > 20000]
    print(price_more_than_20000)
    print('Total Inventory Value:',sum(list_of_mobile_price))
    print('Most Expensive mobile price: ',max(list_of_mobile_price))
mobile_store_management(list_of_mobile_price)

```

Q3. A customer adds products to a shopping cart. Write a function that 
accepts a list of product names and performs the following 
operations: 
• Add a new product.
• Remove a purchased or unwanted product. 
• Display the final cart. 
• Count the total number of products.

```python
product_list = ['Mouse','Laptop','Keyboard','Pen','Jeans']
def shopping_cart(product_list,new_product,remove_item):
    if new_product in product_list:
        print("Product Already in Cart")
    else:
        product_list.append(new_product)
    if remove_item in product_list:
        product_list.remove(remove_item)
    print(product_list)
    print("The Total number of products: ",len(product_list))
shopping_cart(product_list,'Shirt','Keyboard')

```

Q 20. A customer support team stores complaint IDs and their statuses 
in a dictionary. Write a function to: 
• Add a new complaint. 
• Update the status of an existing complaint. 
• Count pending and resolved complaints. 
• Display all complaint records.

```python
def complaint_tracker(complaints, new_id, new_status,
                      update_id, updated_status):

    complaints[new_id] = new_status

    if update_id in complaints:
        complaints[update_id] = updated_status
    else:
        print("Complaint ID not found")

    pending = sum(
        1 for status in complaints.values()
        if status.lower() == "pending"
    )

    resolved = sum(
        1 for status in complaints.values()
        if status.lower() == "resolved"
    )

    print("Complaint records:", complaints)
    print("Pending complaints:", pending)
    print("Resolved complaints:", resolved)


complaint_tracker(
    {1: "Pending", 2: "Resolved", 3: "Pending"},
    4,
    "Pending",
    1,
    "Resolved"
)
```

Q19. A business records the number of units sold for each product in a 
dictionary. Write a function to calculate total units sold, identify 
the best-selling product, and display products with sales below 10 
units.

```python
def analyze_sales(sales):

    print("Total units sold:", sum(sales.values()))

    best_product = max(sales, key=sales.get)
    print("Best-selling product:",
          best_product, "-", sales[best_product], "units")

    print("Products with sales below 10 units:",
          [product for product, units in sales.items()
           if units < 10])


analyze_sales({
    "Laptop": 15,
    "Mouse": 35,
    "Keyboard": 8,
    "Webcam": 6,
    "Headphones": 12
})
```

Q4. A company stores employee salaries in a list. Write a function to 
calculate the average salary, display salaries above the average, 
and increase every salary by 10%.

```python
salary_list = [15000,45000,45900,46900,78000,49231]
def employee_salaries(salary_list):
    average = sum(salary_list) / len(salary_list)
    print(f'The average salary of employee: {average:.2f}')
    print('Salary above the average: ',[salary for salary in salary_list if salary > average])
employee_salaries(salary_list)
```
